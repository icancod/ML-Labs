```python
def detect_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    print(f"{column}: {len(outliers)} outliers")
    return outliers
```

```python
detect_outliers('daily_social_media_hours')
detect_outliers('sleep_hours')
detect_outliers('stress_level')
detect_outliers('anxiety_level')
```

Using the IQR method, no outliers were detected in daily social media hours, sleep hours, stress level, or anxiety level. This indicates that all observations lie within statistically acceptable bounds.
The absence of outliers suggests that extreme values in social media usage are not data entry errors, but rather represent realistic behavioural variations among teenagers. For example, higher social media usage may indicate strong digital engagement or addictive behaviour, while lower usage may reflect limited online activity. Therefore, these extreme values should be treated as meaningful behavioural patterns rather than removed from the dataset.
