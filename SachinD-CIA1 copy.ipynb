## Question 3: Missing Values & Imputation

```python
missing_count = original_df.isnull().sum()
missing_percentage = (missing_count / len(original_df)) * 100

missing_df = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print(missing_df)
```

### Imputation Strategy

**Part (b) - Social Media & Sleep Variables:**
- For `daily_social_media_hours, `sleep_hours`, and `screen_time_before_sleep: Use the **median** value to fill blanks. These are straightforward numeric variables where the median is robust and works well.

**Part (c) - Mental Health Variables:**
- For `stress_level`, `anxiety_level`, and `addiction_level`: Use **median** or **K-nearest neighbors (KNN)** imputation. KNN might work better here since teens with similar profiles might have similar mental health patterns.

**Part (d) - Missing Data Mechanism:**
- **MAR (Missing At Random):** Some anxiety responses might be missing because anxious teens use certain platforms more, or their gender influences help-seeking behavior.
- **MNAR (Missing Not At Random):** Teens with very high anxiety might deliberately skip the anxiety question, so the missingness itself is related to the unobserved anxiety level.

With depression, there's a slightly clearer story but still nothing huge. Stress and anxiety go up when depression is present, and sleep goes down—that makes sense intuitively. But even these relationships aren't super strong (roughly 0.17 and −0.20), so depression isn't driven by just one or two factors.

**A note on categories:** We can't just throw gender or platform into a correlation matrix because correlation needs numbers that actually *mean* something mathematically. Categories are labels, not quantities. You'd need to code them as numbers first (like 1 for male, 2 for female), and even then you have to be careful because those numbers would be somewhat arbitrary.

## Question 6: Outlier Detection

```python
def detect_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    print(f"{column}: {len(outliers)} outliers")
    return outliers
```

```python
detect_outliers('daily_social_media_hours')
detect_outliers('sleep_hours')
detect_outliers('stress_level')
detect_outliers('anxiety_level')
```

**IQR Outlier Detection Results:**

No outliers were found using the standard IQR method (Q1 - 1.5×IQR to Q3 + 1.5×IQR). All values sit comfortably within the boundaries. This makes sense because the data seems to be synthetic or carefully controlled—there aren't any wild spikes that stick out as obvious errors. Even the extreme values (like 8 hours of social media) are realistic for a teenager, so there's nothing to flag or remove.
